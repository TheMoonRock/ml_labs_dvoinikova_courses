Проверить эту гипотезу замерив время выполняения в задаче токенизации и удаления пунктуации.

```python
text = "Лондон — столица и крупнейший город Англии и Великобритании. Административно образует регион Англии Большой Лондон, разделённый на 33 самоуправляемых территории (см. боро Лондона). Население города в 2023 году составляло 9 648 110 человек. Образует агломерацию Большой Лондон и более обширный метрополитенский район. Расположен на юго-востоке острова Великобритания, на равнине в устье Темзы, вблизи Северного моря. Главный политический, экономический и культурный центр Соединённого Королевства. Относится к глобальным городам высшего ранга и ведущим мировым финансовым центрам. Экономика Лондона составляет пятую часть экономики страны. По объёму ВРП занимает 5—6-е место среди региональных агломераций мира. Один из привлекательнейших городов для инвестиций, в его юрисдикции зарегистрировано наибольшее число международных торговых компаний и лиц со сверхкрупным чистым капиталом среди всех городов мира. Лондон основан римлянами вскоре после их вторжения на Британские острова в 43 году н. э. Приблизительно с 100 года — столица римской Британии, с XI—XII столетий — Англии, с 1707 года — Великобритании, с XVI по XX век — Британской империи. С 1825 по 1925 год Лондон был крупнейшим городом мира. Исторический центр города, образованный районами Вестминстер и Сити, сложился в викторианскую эпоху. Среди немногих построек, уцелевших после пожара 1666 года, — средневековая цитадель Тауэр. В 2020 году, в ознаменование 80-летия «Речи 18 июня» генерала де Голля, Лондон отмечен орденом Почётного легиона."
```

```python
import time
import re
start = time.time()

print(re.split('[.?!-,—]', text))


end = time.time()
regex_time = end - start
print(regex_time)
```

```python
import nltk
# Скачиваем необходимые пакеты
nltk.download('punkt')  # для токенизации
nltk.download('punkt_tab')  # новая версия (может потребоваться)
nltk.download('stopwords')  # для стоп-слов (опционально)
```

```python
from string import punctuation
from nltk.tokenize import word_tokenize
start = time.time()
punctuations = list(punctuation)

words_without_punct = [i for i in word_tokenize(text) if (i not in punctuations)]

print(words_without_punct)
end = time.time()
lib_time = end - start
print(lib_time)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

exp_names = ['regex_time', 'lib_time']
exp_times = [regex_time, lib_time]
#bar_labels = ['red', 'blue', '_red', 'orange']
#bar_colors = ['tab:red', 'tab:blue', 'tab:red', 'tab:orange']

ax.bar(exp_names, exp_times)

ax.set_ylabel('seconds')
ax.set_title('experiments difference of time')
ax.legend(title='time')

plt.show()
```

```python
print(f'regex_time: {regex_time}')
print(f'lib_time: {lib_time}')
```

## 3. Проверить пропущенные значения. Создать колонку с длиной текстов. Вычислить среднюю, медиану и моду.

```python
result['length_of_text'] = result['ttext'].str.len()
result
```

```python
print(result.isna().sum())
```

Средняя – среднее арифметическое всего вариационного ряда. 

Медиана – это значение признака, находящееся в середине ранжированной (упорядоченной по возрастанию или убыванию) совокупности.

Мода – значение признака, наиболее часто встречающееся в выборке.

```python
length_of_text_mean = result['length_of_text'].mean()
print(f'среднее: {length_of_text_mean}')
```

```python
length_of_text_median = result['length_of_text'].median()
print(f'медиана: {length_of_text_median}')
```

```python
length_of_text_mode = result['length_of_text'].mode()[0]
print(f'мода: {length_of_text_mode}')
```

```python
result['length_of_text'].describe()
```

В выводе метода .describe() слова «медиана» действительно нет, но медиана там скрывается под обозначением 50%.
50% — это и есть медиана (второй квартиль). Это значение, которое делит вашу выборку ровно пополам: 50% текстов имеют длину меньше 74 символов, а другие 50% — больше.

* count (10000.0) — количество заполненных строк.
* mean (80.72) — среднее арифметическое значение.
* std (27.86) — стандартное отклонение (насколько числа разбросаны относительно среднего).
* min (41.0) — минимум (самый короткий текст).
* 25% (58.0) — первый квартиль (25% текстов короче этого значения).
* 50% (74.0) — медиана (ровно половина текстов короче 74).
* 75% (100.0) — третий квартиль (75% текстов короче этого значения).
* max (164.0) — максимум (самый длинный текст).

Единственное, чего действительно нет в этом списке — это моды (самого частого значения). Её всегда нужно считать отдельно через .mode().

```python
# Группируем по типу текста и выводим метрики для колонки с длиной
result.groupby('ttype')['length_of_text'].describe()
```

Тут можно заметить что среднее и медиана длины положительных текстов больше чем у отрицательных

## 4. Выполнить визуальный анализ текстового корпуса Рубцовой. Построить минимум 5 графиков, минимум по 1 из каждой библиотеки.

Обязательно нужно использовать все три библиотеки.
1. matplotlib
2. seaborn
3. plotly

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly
import plotly.graph_objs as go
import plotly.express as px
from plotly.subplots import make_subplots
import numpy as np
```

```python
# Устанавливаем стиль
sns.set_theme(style="ticks", palette="pastel")

# Создаем график
plt.figure(figsize=(10, 6))
sns.boxplot(x="ttype", y="length_of_text",
            data=result,
            palette=["m", "g"])  # m - magenta для отрицательных, g - green для положительных

# Настраиваем подписи
plt.xlabel('Тип текста', fontsize=12)
plt.ylabel('Длина текста', fontsize=12)
plt.title('Распределение длины текста по типам', fontsize=14)

# Меняем подписи на оси x для удобства
plt.xticks([0, 1], ['Отрицательные (-1)', 'Положительные (1)'])

# Убираем лишние рамки
sns.despine(offset=10, trim=True)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Создаем колонку с квартилями длины текста
result['length_quartile'] = pd.qcut(result['length_of_text'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'])

# Группируем по типу текста и квартилю, считаем среднюю длину
grouped_stats = result.groupby(['ttype', 'length_quartile'])['length_of_text'].mean().reset_index()

# Создаем сводную таблицу для тепловой карты
pivot_table = grouped_stats.pivot(index='length_quartile', 
                                   columns='ttype', 
                                   values='length_of_text')

# Переименовываем колонки для понятности
pivot_table.columns = ['Отрицательные (-1)', 'Положительные (1)']
pivot_table.index.name = 'Квартиль длины'

print("Средняя длина текста по квартилям и типам:")
print(pivot_table)
print("\n")

# Визуализация в виде тепловой карты
plt.figure(figsize=(10, 6))
sns.heatmap(pivot_table, annot=True, fmt='.1f', cmap='coolwarm', 
            cbar_kws={'label': 'Средняя длина текста'})
plt.title('Средняя длина текста по квартилям и типам', fontsize=14)
plt.xlabel('Тип текста', fontsize=12)
plt.ylabel('Квартиль длины', fontsize=12)
plt.tight_layout()
plt.show()

# Или альтернативно - столбчатая диаграмма сгруппированная
fig, ax = plt.subplots(figsize=(10, 6))
pivot_table.plot(kind='bar', ax=ax, color=['magenta', 'green'])
ax.set_xlabel('Квартиль длины', fontsize=12)
ax.set_ylabel('Средняя длина текста', fontsize=12)
ax.set_title('Средняя длина текста по квартилям и типам', fontsize=14)
ax.legend(['Отрицательные (-1)', 'Положительные (1)'])
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
